```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data= {
    "height":[185, 170, 168, 179, 182, 188, 180, 183, 180, 180, 177],
    "weight": [72, 56,60, 68, 72, 77, 70, 84, 88, 67, 76]
}

df= pd.DataFrame(data)
print(df)
```

```python
#Plot dataset
plt.figure(figsize=(7,6))

plt.scatter(df["height"], df["weight"], s=120)

for i in range(len(df)):
    plt.text(df["height"][i]+0.2,
             df["weight"][i]+0.2,
             str(i+1))

plt.xlabel("height")
plt.ylabel("weight")
plt.title("Original Dataset")
plt.grid(True)

plt.show()
```

```python
initial_centroids= np.array([
    [185, 72],
    [170, 56]
])
```

```python
kmeans= KMeans(
    n_clusters=2,
    init= initial_centroids,
    n_init=1,
    random_state=42
)

kmeans.fit(df)
```

```python
#Plot cluster
plt.figure(figsize=(7,6))

colors = ['red','blue']

for i in range(2):
    cluster = df[df["cluster"]==i]

    plt.scatter(
        cluster["height"],
        cluster["weight"],
        s=120,
        color=colors[i],
        label=f"Cluster {i+1}"
    )

plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    color="black",
    marker="X",
    s=300,
    label="Centroids"
)

plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("K-Means Clustering")
plt.legend()
plt.grid(True)

plt.show()
```
